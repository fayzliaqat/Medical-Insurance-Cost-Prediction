## 1. Load Dataset

```python
import pandas as pd

df = pd.read_csv('../data/insurance.csv')
print('Dataset loaded successfully.')
print('Initial shape:', df.shape)
```

## 2. Missing-Value Analysis

```python
print('Missing values per column:')
print(df.isnull().sum())
print()
print('Total missing values:', df.isnull().sum().sum())
```

> **Result:** No missing values found in any column.  
> No missing-value treatment was required.

## 3. Duplicate Analysis

```python
print('Duplicate rows before removal:', df.duplicated().sum())
print('Row count before:', df.shape[0])
```

```python
df = df.drop_duplicates().reset_index(drop=True)
print('Duplicate rows after removal:', df.duplicated().sum())
print('Row count after:', df.shape[0])
```

> **Result:** 1,435 duplicate rows were found and removed.  
> Dataset reduced from **2,772 rows** to **1,337 rows**.

## 4. Data-Type Checks

```python
df.dtypes
```

> **Result:** All numeric columns (`age`, `bmi`, `children`, `charges`) have correct numeric dtypes.  
> Categorical columns (`sex`, `smoker`, `region`) are stored as `object` — appropriate before encoding.  
> No dtype conversions were necessary.

## 5. Categorical & Numeric Value Checks

```python
print('sex unique values   :', df['sex'].unique())
print('smoker unique values:', df['smoker'].unique())
print('region unique values:', df['region'].unique())
```

```python
print('age    min:', df['age'].min())
print('bmi    min:', df['bmi'].min())
print('children min:', df['children'].min())
print('charges min:', df['charges'].min())
```

> **Result:** All categorical values are valid and consistent.  
> All numeric minimums are non-negative and within realistic ranges.  
> No invalid or inconsistent values found.

## 6. Before Preprocessing — Categorical Columns (Original State)

```python
print('Categorical columns before encoding:')
df[['sex', 'smoker', 'region']].head()
```

## 7. Categorical Encoding

Encode `sex`, `smoker`, and `region` using one-hot encoding (`pd.get_dummies`).  
`drop_first=True` removes one redundant dummy column per variable to avoid multicollinearity — suitable for linear regression.  

Columns created:
- `sex` → `sex_male` (1 = male, 0 = female)
- `smoker` → `smoker_yes` (1 = smoker, 0 = non-smoker)
- `region` → `region_northwest`, `region_southeast`, `region_southwest` (northeast is the dropped baseline)


```python
df = pd.get_dummies(df, columns=['sex', 'smoker', 'region'], drop_first=True)
print('Encoding complete.')
print('New shape:', df.shape)
```
